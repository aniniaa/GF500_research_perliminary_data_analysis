```python
import pandas as pd
import re

def normalize_company_name(name):
    """
    Normalize company name for matching
    """
    if pd.isna(name):
        return ''
    
    # Convert to uppercase
    name = str(name).upper().strip()
    
    # Remove common legal suffixes
    suffixes = [
        r'\bLTD\.?$', r'\bLIMITED$', r'\bINC\.?$', r'\bCORP\.?$', 
        r'\bCORPORATION$', r'\bPLC\.?$', r'\bLLC\.?$', r'\bLLP\.?$',
        r'\bGROUP$', r'\bHOLDINGS?$', r'\bCOMPANY$', r'\bCO\.?$',
        r'\bSA$', r'\bAG$', r'\bGMBH$', r'\bNV$', r'\bBV$',
        r'\bPTE\.?$', r'\bPVT\.?$', r'\bLTDA?\.?$'
    ]
    
    for suffix in suffixes:
        name = re.sub(suffix, '', name).strip()
    
    # Remove special characters but keep spaces
    name = re.sub(r'[^A-Z0-9\s]', ' ', name)
    
    # Remove extra whitespace
    name = re.sub(r'\s+', ' ', name).strip()
    
    return name

# Load your customer database
# Replace 'customer_db.csv' with your actual file path
df = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\All Accounts Report-2026-02-05.xlsx')

# Create normalized name column
df['Normalized_Name'] = df['Account Name'].apply(normalize_company_name)

# Find duplicates based on normalized names
duplicates = df[df.duplicated(subset=['Normalized_Name'], keep=False)]

# Group by normalized name and sort
duplicate_groups = duplicates.groupby('Normalized_Name')

# Display results
print(f"Total unique companies in database: {df['Normalized_Name'].nunique()}")
print(f"Total records in database: {len(df)}")
print(f"Companies with duplicates: {duplicates['Normalized_Name'].nunique()}")
print(f"Total duplicate records: {len(duplicates)}")
print("\n" + "="*100 + "\n")

# Show each duplicate group with full details
for norm_name, group in duplicate_groups:
    print(f"NORMALIZED NAME: {norm_name}")
    print(f"Number of records: {len(group)}")
    print("-" * 100)
    
    # Display all columns for each record in the group
    for idx, row in group.iterrows():
        print(f"\nRecord {idx + 1}:")
        print(f"  Account ID: {row['AccountID_18']}")
        print(f"  Original Name: {row['Account Name']}")
        print(f"  Type: {row['Type']}")
        print(f"  Location: {row['Billing City']}, {row['Billing State/Province']}, {row['Billing Country']}")
        print(f"  Industry: {row['Industry']} - {row['Industry: Sub-category']}")
        print(f"  Account Owner: {row['Account Owner']}")
        print(f"  Net Zero Target Year: {row['Net Zero Target Year']}")
        print(f"  RE100 Member: {row['RE100 Member']}")
        print(f"  Website: {row['Website']}")
    
    print("\n" + "="*100 + "\n")

# Optional: Save duplicate groups to Excel for easier review
with pd.ExcelWriter('customer_duplicates.xlsx') as writer:
    for norm_name, group in duplicate_groups:
        # Excel sheet names have 31 char limit
        sheet_name = norm_name[:31] if len(norm_name) > 31 else norm_name
        group.to_excel(writer, sheet_name=sheet_name, index=False)

print("Duplicate groups saved to 'customer_duplicates.xlsx'")

# Also create a summary view
summary = duplicates.groupby('Normalized_Name').agg({
    'AccountID_18': 'count',
    'Account Name': lambda x: ' | '.join(x.unique()),
    'Type': lambda x: ' | '.join(x.unique()),
    'Billing Country': lambda x: ' | '.join(x.unique())
}).rename(columns={'AccountID_18': 'Count'})

summary.to_excel('duplicates_summary.xlsx')
print("Summary saved to 'duplicates_summary.xlsx'")
```

```python
import pandas as pd
import re

def normalize_company_name(name):
    if pd.isna(name) or str(name).strip() in ['', '-', 'nan']:
        return None
    
    name = str(name).upper().strip()
    
    suffixes = [
        r'\bLTD\.?$', r'\bLIMITED$', r'\bINC\.?$', r'\bCORP\.?$', 
        r'\bCORPORATION$', r'\bPLC\.?$', r'\bLLC\.?$', r'\bLLP\.?$',
        r'\bGROUP$', r'\bHOLDINGS?$', r'\bCOMPANY$', r'\bCO\.?$',
        r'\bSA$', r'\bAG$', r'\bGMBH$', r'\bNV$', r'\bBV$',
        r'\bPTE\.?$', r'\bPVT\.?$', r'\bLTDA?\.?$'
    ]
    
    for suffix in suffixes:
        name = re.sub(suffix, '', name).strip()
    
    name = re.sub(r'[^A-Z0-9\s]', ' ', name)
    name = re.sub(r'\s+', ' ', name).strip()
    
    if len(name) < 2:
        return None
    
    return name

# Load databases
customer_db = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\All Accounts Report-2026-02-05.xlsx')
gf500 = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\comparative analysis\FG500_DataCollection_FY26_v4.xlsm', header=2)

# Normalize names
customer_db['Normalized_Name'] = customer_db['Account Name'].apply(normalize_company_name)
gf500['Normalized_Name'] = gf500['Company name'].apply(normalize_company_name)

# Filter valid names
customer_db_valid = customer_db[customer_db['Normalized_Name'].notna()]
gf500_valid = gf500[gf500['Normalized_Name'].notna()]

# Match customers with GF500
matched = customer_db_valid.merge(
    gf500_valid,
    on='Normalized_Name',
    how='inner',
    suffixes=('_Customer', '_GF500')
)

print(f"Total customer records matched with GF500: {len(matched)}")
print(f"Unique GF500 companies: {matched['Normalized_Name'].nunique()}")

# Split by type
current_clients = matched[matched['Type'].isin(['Existing Client', 'New Client'])].copy()
prospects = matched[matched['Type'] == 'Prospect'].copy()
lapsed_clients = matched[matched['Type'].isin(['Lapsed client', 'Ex-client'])].copy()

print(f"\nCurrent Clients: {len(current_clients)} records, {current_clients['Normalized_Name'].nunique()} unique companies")
print(f"Prospects: {len(prospects)} records, {prospects['Normalized_Name'].nunique()} unique companies")
print(f"Lapsed Clients: {len(lapsed_clients)} records, {lapsed_clients['Normalized_Name'].nunique()} unique companies")

# Select key columns from customer DB
customer_columns = [
    'AccountID_18',
    'Account Name',
    'Type',
    'Billing City',
    'Billing State/Province',
    'Billing Country',
    'Industry',
    'Industry: Sub-category',
    'Account Owner',
    'Website'
]

# Select key columns from GF500
gf500_columns = [
    'Company name',
    'Net Zero Interim Target Dates',
    'Net Zero Target Status',
    'Net Zero target set',
    'Net Zero Target Year',
    'RE100 Target Year',
    'RE100 Member',
    'Target Type',
    'ZI Website',
    'RE100',
    'SBTi NT',
    'Net Zero',
    'Carbon Neutral'
]

# Create tracker with selected columns
def create_tracker_df(df, customer_cols, gf500_cols):
    tracker = df.copy()
    
    # Rename GF500 columns to avoid confusion
    rename_dict = {col: f'GF500_{col}' for col in gf500_cols if col != 'Company name'}
    tracker = tracker.rename(columns=rename_dict)
    
    # Select and reorder columns
    selected_cols = customer_cols + ['GF500_' + col if col != 'Company name' else col for col in gf500_cols]
    selected_cols = [col for col in selected_cols if col in tracker.columns]
    
    return tracker[selected_cols]

# Create trackers
clients_tracker = create_tracker_df(current_clients, customer_columns, gf500_columns)
prospects_tracker = create_tracker_df(prospects, customer_columns, gf500_columns)
lapsed_tracker = create_tracker_df(lapsed_clients, customer_columns, gf500_columns)

# Sort by GF500 company name
clients_tracker = clients_tracker.sort_values('Company name')
prospects_tracker = prospects_tracker.sort_values('Company name')
lapsed_tracker = lapsed_tracker.sort_values('Company name')

# Export to Excel
with pd.ExcelWriter('/mnt/user-data/outputs/GF500_Customer_Tracker.xlsx', engine='openpyxl') as writer:
    # Current Clients
    clients_tracker.to_excel(writer, sheet_name='Current Clients', index=False)
    
    # Prospects
    prospects_tracker.to_excel(writer, sheet_name='Prospects', index=False)
    
    # Lapsed Clients
    lapsed_tracker.to_excel(writer, sheet_name='Lapsed Clients', index=False)
    
    # Summary sheet
    summary_data = {
        'Category': ['Current Clients', 'Prospects', 'Lapsed Clients', 'TOTAL'],
        'Number of Records': [len(clients_tracker), len(prospects_tracker), len(lapsed_tracker), 
                              len(clients_tracker) + len(prospects_tracker) + len(lapsed_tracker)],
        'Unique GF500 Companies': [clients_tracker['Company name'].nunique(), 
                                   prospects_tracker['Company name'].nunique(),
                                   lapsed_tracker['Company name'].nunique(),
                                   matched['Company name'].nunique()]
    }
    summary = pd.DataFrame(summary_data)
    summary.to_excel(writer, sheet_name='Summary', index=False)
    
    # Auto-adjust column widths
    for sheet_name in writer.sheets:
        worksheet = writer.sheets[sheet_name]
        for column in worksheet.columns:
            max_length = 0
            column_letter = column[0].column_letter
            for cell in column:
                try:
                    if len(str(cell.value)) > max_length:
                        max_length = len(str(cell.value))
                except:
                    pass
            adjusted_width = min(max_length + 2, 50)
            worksheet.column_dimensions[column_letter].width = adjusted_width

print("\n" + "="*100)
print("GF500 Customer Tracker created successfully!")
print("="*100)

print("\n\nTRACKER CONTENTS:")
print(f"✓ Current Clients: {len(clients_tracker)} records")
print(f"✓ Prospects: {len(prospects_tracker)} records")
print(f"✓ Lapsed Clients: {len(lapsed_tracker)} records")
print(f"\nEach sheet includes:")
print("  - Customer info (Account ID, Name, Type, Location, Industry, Owner, Website)")
print("  - GF500 climate targets (Net Zero, RE100, SBTi, Carbon Neutral, Target Years)")
print("\nFile saved to: 'GF500_Customer_Tracker.xlsx'")
```
