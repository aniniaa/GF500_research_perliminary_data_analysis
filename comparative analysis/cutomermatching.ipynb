```python
import pandas as pd
import re

def normalize_company_name(name):
    """
    Normalize company name for matching
    """
    if pd.isna(name):
        return ''
    
    # Convert to uppercase
    name = str(name).upper().strip()
    
    # Remove common legal suffixes
    suffixes = [
        r'\bLTD\.?$', r'\bLIMITED$', r'\bINC\.?$', r'\bCORP\.?$', 
        r'\bCORPORATION$', r'\bPLC\.?$', r'\bLLC\.?$', r'\bLLP\.?$',
        r'\bGROUP$', r'\bHOLDINGS?$', r'\bCOMPANY$', r'\bCO\.?$',
        r'\bSA$', r'\bAG$', r'\bGMBH$', r'\bNV$', r'\bBV$',
        r'\bPTE\.?$', r'\bPVT\.?$', r'\bLTDA?\.?$'
    ]
    
    for suffix in suffixes:
        name = re.sub(suffix, '', name).strip()
    
    # Remove special characters but keep spaces
    name = re.sub(r'[^A-Z0-9\s]', ' ', name)
    
    # Remove extra whitespace
    name = re.sub(r'\s+', ' ', name).strip()
    
    return name

# Load your customer database
# Replace 'customer_db.csv' with your actual file path
df = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\All Accounts Report-2026-02-05.xlsx')

# Create normalized name column
df['Normalized_Name'] = df['Account Name'].apply(normalize_company_name)

# Find duplicates based on normalized names
duplicates = df[df.duplicated(subset=['Normalized_Name'], keep=False)]

# Group by normalized name and sort
duplicate_groups = duplicates.groupby('Normalized_Name')

# Display results
print(f"Total unique companies in database: {df['Normalized_Name'].nunique()}")
print(f"Total records in database: {len(df)}")
print(f"Companies with duplicates: {duplicates['Normalized_Name'].nunique()}")
print(f"Total duplicate records: {len(duplicates)}")
print("\n" + "="*100 + "\n")

# Show each duplicate group with full details
for norm_name, group in duplicate_groups:
    print(f"NORMALIZED NAME: {norm_name}")
    print(f"Number of records: {len(group)}")
    print("-" * 100)
    
    # Display all columns for each record in the group
    for idx, row in group.iterrows():
        print(f"\nRecord {idx + 1}:")
        print(f"  Account ID: {row['AccountID_18']}")
        print(f"  Original Name: {row['Account Name']}")
        print(f"  Type: {row['Type']}")
        print(f"  Location: {row['Billing City']}, {row['Billing State/Province']}, {row['Billing Country']}")
        print(f"  Industry: {row['Industry']} - {row['Industry: Sub-category']}")
        print(f"  Account Owner: {row['Account Owner']}")
        print(f"  Net Zero Target Year: {row['Net Zero Target Year']}")
        print(f"  RE100 Member: {row['RE100 Member']}")
        print(f"  Website: {row['Website']}")
    
    print("\n" + "="*100 + "\n")

# Optional: Save duplicate groups to Excel for easier review
with pd.ExcelWriter('customer_duplicates.xlsx') as writer:
    for norm_name, group in duplicate_groups:
        # Excel sheet names have 31 char limit
        sheet_name = norm_name[:31] if len(norm_name) > 31 else norm_name
        group.to_excel(writer, sheet_name=sheet_name, index=False)

print("Duplicate groups saved to 'customer_duplicates.xlsx'")

# Also create a summary view
summary = duplicates.groupby('Normalized_Name').agg({
    'AccountID_18': 'count',
    'Account Name': lambda x: ' | '.join(x.unique()),
    'Type': lambda x: ' | '.join(x.unique()),
    'Billing Country': lambda x: ' | '.join(x.unique())
}).rename(columns={'AccountID_18': 'Count'})

summary.to_excel('duplicates_summary.xlsx')
print("Summary saved to 'duplicates_summary.xlsx'")
```

```python
import pandas as pd
import re

def normalize_company_name(name):
    """
    Normalize company name for matching
    """
    if pd.isna(name):
        return ''
    
    # Convert to uppercase
    name = str(name).upper().strip()
    
    # Remove common legal suffixes
    suffixes = [
        r'\bLTD\.?$', r'\bLIMITED$', r'\bINC\.?$', r'\bCORP\.?$', 
        r'\bCORPORATION$', r'\bPLC\.?$', r'\bLLC\.?$', r'\bLLP\.?$',
        r'\bGROUP$', r'\bHOLDINGS?$', r'\bCOMPANY$', r'\bCO\.?$',
        r'\bSA$', r'\bAG$', r'\bGMBH$', r'\bNV$', r'\bBV$',
        r'\bPTE\.?$', r'\bPVT\.?$', r'\bLTDA?\.?$'
    ]
    
    for suffix in suffixes:
        name = re.sub(suffix, '', name).strip()
    
    # Remove special characters but keep spaces
    name = re.sub(r'[^A-Z0-9\s]', ' ', name)
    
    # Remove extra whitespace
    name = re.sub(r'\s+', ' ', name).strip()
    
    return name

# Load databases
customer_db = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\All Accounts Report-2026-02-05.xlsx')
gf500 = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\comparative analysis\FG500_DataCollection_FY26_v4.xlsm', header=2)



# Normalize names in both databases
customer_db['Normalized_Name'] = customer_db['Account Name'].apply(normalize_company_name)
gf500['Normalized_Name'] = gf500['Company name'].apply(normalize_company_name)

print(f"Customer DB records: {len(customer_db)}")
print(f"GF500 records: {len(gf500)}")
print("\n" + "="*100 + "\n")

# 1. EXACT MATCHES
exact_matches = customer_db.merge(
    gf500[['Company name', 'Normalized_Name']], 
    on='Normalized_Name', 
    how='inner',
    suffixes=('_Customer', '_GF500')
)

print(f"EXACT MATCHES: {len(exact_matches)}")
print("-" * 100)

# Check for duplicates in exact matches (same GF500 company matched to multiple customer records)
gf500_duplicates = exact_matches.groupby('Normalized_Name').filter(lambda x: len(x) > 1)

if len(gf500_duplicates) > 0:
    print(f"\nGF500 companies with MULTIPLE customer records: {gf500_duplicates['Normalized_Name'].nunique()}\n")
    
    for norm_name, group in gf500_duplicates.groupby('Normalized_Name'):
        print(f"GF500 COMPANY: {group.iloc[0]['Company name']}")
        print(f"Normalized: {norm_name}")
        print(f"Matched to {len(group)} customer records:")
        print("-" * 100)
        
        for idx, row in group.iterrows():
            print(f"\n  Customer Record:")
            print(f"    Account ID: {row['AccountID_18']}")
            print(f"    Account Name: {row['Account Name']}")
            print(f"    Type: {row['Type']}")
            print(f"    Location: {row['Billing City']}, {row['Billing Country']}")
            print(f"    Industry: {row['Industry']}")
        
        print("\n" + "="*100 + "\n")
else:
    print("No duplicate matches found (1-to-1 mapping)")

# 2. SUBSTRING MATCHING (one name contains the other)
unmatched_customers = customer_db[~customer_db['Normalized_Name'].isin(exact_matches['Normalized_Name'])]
unmatched_gf500 = gf500[~gf500['Normalized_Name'].isin(exact_matches['Normalized_Name'])]

print(f"\nUNMATCHED CUSTOMERS: {len(unmatched_customers)}")
print(f"UNMATCHED GF500: {len(unmatched_gf500)}")

print("\n" + "="*100)
print("SUBSTRING MATCHES (shorter name fully contained in longer name)")
print("="*100 + "\n")

substring_results = []
for idx, customer_row in unmatched_customers.iterrows():
    customer_norm = customer_row['Normalized_Name']
    if not customer_norm or len(customer_norm) < 3:  # Skip empty or very short names
        continue
    
    for gf500_idx, gf500_row in unmatched_gf500.iterrows():
        gf500_norm = gf500_row['Normalized_Name']
        if not gf500_norm or len(gf500_norm) < 3:
            continue
        
        # Check if one is contained in the other
        if customer_norm in gf500_norm or gf500_norm in customer_norm:
            substring_results.append({
                'Customer_AccountID': customer_row['AccountID_18'],
                'Customer_Name': customer_row['Account Name'],
                'Customer_Normalized': customer_norm,
                'GF500_Name': gf500_row['Company name'],
                'GF500_Normalized': gf500_norm,
                'Match_Type': 'Customer shorter' if customer_norm in gf500_norm else 'GF500 shorter'
            })

if substring_results:
    substring_df = pd.DataFrame(substring_results)
    print(substring_df.to_string(index=False))
    print(f"\nTotal substring matches found: {len(substring_results)}")
else:
    print("No substring matches found")

# 3. EXPORT RESULTS
with pd.ExcelWriter('gf500_matching_results.xlsx') as writer:
    # Exact matches
    exact_matches.to_excel(writer, sheet_name='Exact Matches', index=False)
    
    # Duplicates (GF500 companies with multiple customer records)
    if len(gf500_duplicates) > 0:
        gf500_duplicates.to_excel(writer, sheet_name='Duplicate Matches', index=False)
    
    # Substring matches
    if substring_results:
        substring_df.to_excel(writer, sheet_name='Substring Matches', index=False)
    
    # Unmatched customers
    unmatched_customers.to_excel(writer, sheet_name='Unmatched Customers', index=False)
    
    # Unmatched GF500
    unmatched_gf500.to_excel(writer, sheet_name='Unmatched GF500', index=False)

print("\n" + "="*100)
print("Results saved to 'gf500_matching_results.xlsx'")
print("="*100)

# Summary statistics
print("\n\nSUMMARY:")
print(f"Total customer records: {len(customer_db)}")
print(f"Total GF500 companies: {len(gf500)}")
print(f"Exact matches: {len(exact_matches)}")
print(f"  - GF500 companies with duplicates: {gf500_duplicates['Normalized_Name'].nunique() if len(gf500_duplicates) > 0 else 0}")
print(f"Substring matches: {len(substring_results)}")
print(f"Unmatched customers: {len(unmatched_customers) - len(substring_results)}")
print(f"Unmatched GF500: {len(unmatched_gf500)}")
```

```python
import pandas as pd
import re

def normalize_company_name(name):
    """
    Basic normalization - uppercase and remove common suffixes
    """
    if pd.isna(name) or str(name).strip() in ['', '-', 'nan']:
        return None
    
    # Convert to uppercase
    name = str(name).upper().strip()
    
    # Remove common legal suffixes
    suffixes = [
        r'\bLTD\.?$', r'\bLIMITED$', r'\bINC\.?$', r'\bCORP\.?$', 
        r'\bCORPORATION$', r'\bPLC\.?$', r'\bLLC\.?$', r'\bLLP\.?$',
        r'\bGROUP$', r'\bHOLDINGS?$', r'\bCOMPANY$', r'\bCO\.?$',
        r'\bSA$', r'\bAG$', r'\bGMBH$', r'\bNV$', r'\bBV$',
        r'\bPTE\.?$', r'\bPVT\.?$', r'\bLTDA?\.?$'
    ]
    
    for suffix in suffixes:
        name = re.sub(suffix, '', name).strip()
    
    # Remove special characters but keep spaces
    name = re.sub(r'[^A-Z0-9\s]', ' ', name)
    
    # Remove extra whitespace
    name = re.sub(r'\s+', ' ', name).strip()
    
    # Return None if too short
    if len(name) < 2:
        return None
    
    return name


# Load databases
customer_db = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\All Accounts Report-2026-02-05.xlsx')
gf500 = pd.read_excel(r'C:\Users\AniyaBagheri\GF500_research_perliminary_data_analysis\comparative analysis\FG500_DataCollection_FY26_v4.xlsm', header=2)

print(f"Total customer records: {len(customer_db)}")
print(f"Total GF500 records: {len(gf500)}")

# STEP 1: Filter to CURRENT CLIENTS ONLY
client_types = ['Existing Client', 'New Client']
current_clients = customer_db[customer_db['Type'].isin(client_types)].copy()

print(f"\nCurrent clients only (Existing Client + New Client): {len(current_clients)}")
print("\n" + "="*100 + "\n")

# STEP 2: Normalize names
current_clients['Normalized_Name'] = current_clients['Account Name'].apply(normalize_company_name)
gf500['Normalized_Name'] = gf500['Company name'].apply(normalize_company_name)

# Filter out invalid names
current_clients_valid = current_clients[current_clients['Normalized_Name'].notna()].copy()
gf500_valid = gf500[gf500['Normalized_Name'].notna()].copy()

print(f"Current clients with valid names: {len(current_clients_valid)}")
print(f"GF500 with valid names: {len(gf500_valid)}")
print("\n" + "="*100 + "\n")

# STEP 3: EXACT MATCH ONLY
exact_matches = current_clients_valid.merge(
    gf500_valid[['Company name', 'Normalized_Name']], 
    on='Normalized_Name', 
    how='inner',
    suffixes=('_Customer', '_GF500')
)

print(f"EXACT MATCHES:")
print(f"Total matched records: {len(exact_matches)}")
print(f"Unique GF500 companies: {exact_matches['Normalized_Name'].nunique()}")
print(f"Unique customer Account IDs: {exact_matches['AccountID_18'].nunique()}")
print("\n" + "="*100 + "\n")

# Show the matches
print("MATCHED CLIENTS:")
for idx, row in exact_matches.iterrows():
    print(f"{row['Company name']} | Customer: {row['Account Name']} | Type: {row['Type']} | Country: {row['Billing Country']}")

print("\n" + "="*100 + "\n")

# Check for duplicates
duplicates = exact_matches[exact_matches.duplicated(subset=['Normalized_Name'], keep=False)]
if len(duplicates) > 0:
    print(f"DUPLICATES FOUND: {duplicates['Normalized_Name'].nunique()} GF500 companies with multiple client records")
    print("\nDuplicate details:")
    for norm_name, group in duplicates.groupby('Normalized_Name'):
        print(f"\n{group.iloc[0]['Company name']} ({len(group)} records):")
        for idx, row in group.iterrows():
            print(f"  - {row['Account Name']} ({row['Type']}, {row['Billing Country']})")
else:
    print("No duplicates - all 1-to-1 matches")

# Export results
with pd.ExcelWriter('current_clients_gf500_matches.xlsx') as writer:
    exact_matches.to_excel(writer, sheet_name='All Matches', index=False)
    if len(duplicates) > 0:
        duplicates.to_excel(writer, sheet_name='Duplicates', index=False)

print("\n" + "="*100)
print("Results saved to 'current_clients_gf500_matches.xlsx'")
print("="*100)
```
