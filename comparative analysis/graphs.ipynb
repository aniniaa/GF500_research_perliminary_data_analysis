```python
import plotly.graph_objects as go
import numpy as np

# Asia data - adjust these values based on your actual data
region_name = "Asia"
total = 198

# Layer 1: At least 1 target vs No target
at_least_1 = 124
no_target = 74

# Layer 2: Net Zero (from at least 1 target)
net_zero = 54
no_net_zero = 70  # at_least_1 - net_zero

# Layer 3: Carbon Neutral (from Net Zero)
carbon_neutral = 40
no_carbon_neutral = 14  # net_zero - carbon_neutral

# Layer 4: SBTi NT (from Carbon Neutral)
sbti_nt = 30
no_sbti_nt = 10  # carbon_neutral - sbti_nt

# Layer 5: RE100 (from SBTi NT)
re100 = 20
no_re100 = 10  # sbti_nt - re100

# Create figure
fig = go.Figure()

# Layer 1: At least 1 target / No target (innermost ring after hole)
fig.add_trace(go.Pie(
    labels=['At least 1 target', 'No target'],
    values=[at_least_1, no_target],
    hole=0.5,  # Creates the hollow center
    marker=dict(
        colors=['#5BA3C5', '#AAAAAA'],
        line=dict(color='white', width=2)
    ),
    domain={'x': [0.15, 0.85], 'y': [0.15, 0.85]},
    textfont=dict(size=11, color='white'),
    textposition='inside',
    hovertemplate='<b>%{label}</b><br>%{value} companies<br>%{percent}<extra></extra>',
    showlegend=False
))

# Layer 2: Net Zero
fig.add_trace(go.Pie(
    labels=['Net Zero', 'No Net Zero', 'No target (skip)'],
    values=[net_zero, no_net_zero, no_target],
    hole=0.58,
    marker=dict(
        colors=['#00B4D8', '#7BCFE0', 'rgba(0,0,0,0)'],  # Transparent for "No target"
        line=dict(color='white', width=2)
    ),
    domain={'x': [0.12, 0.88], 'y': [0.12, 0.88]},
    textfont=dict(size=10, color='white'),
    textposition='inside',
    hovertemplate='<b>%{label}</b><br>%{value} companies<br>%{percent}<extra></extra>',
    showlegend=False
))

# Layer 3: Carbon Neutral
fig.add_trace(go.Pie(
    labels=['Carbon Neutral', 'No Carbon Neutral', 'Skip'],
    values=[carbon_neutral, no_carbon_neutral, at_least_1 - net_zero + no_target],
    hole=0.66,
    marker=dict(
        colors=['#8B479B', '#A87BB3', 'rgba(0,0,0,0)'],
        line=dict(color='white', width=2)
    ),
    domain={'x': [0.09, 0.91], 'y': [0.09, 0.91]},
    textfont=dict(size=10, color='white'),
    textposition='inside',
    hovertemplate='<b>%{label}</b><br>%{value} companies<br>%{percent}<extra></extra>',
    showlegend=False
))

# Layer 4: SBTi NT
fig.add_trace(go.Pie(
    labels=['SBTi NT', 'No SBTi NT', 'Skip'],
    values=[sbti_nt, no_sbti_nt, total - carbon_neutral],
    hole=0.74,
    marker=dict(
        colors=['#4D58A6', '#7885BA', 'rgba(0,0,0,0)'],
        line=dict(color='white', width=2)
    ),
    domain={'x': [0.06, 0.94], 'y': [0.06, 0.94]},
    textfont=dict(size=10, color='white'),
    textposition='inside',
    hovertemplate='<b>%{label}</b><br>%{value} companies<br>%{percent}<extra></extra>',
    showlegend=False
))

# Layer 5: RE100 (outermost)
fig.add_trace(go.Pie(
    labels=['RE100', 'No RE100', 'Skip'],
    values=[re100, no_re100, total - sbti_nt],
    hole=0.82,
    marker=dict(
        colors=['#9FA8DA', '#B9C1E4', 'rgba(0,0,0,0)'],
        line=dict(color='white', width=2)
    ),
    domain={'x': [0.03, 0.97], 'y': [0.03, 0.97]},
    textfont=dict(size=10, color='white'),
    textposition='inside',
    hovertemplate='<b>%{label}</b><br>%{value} companies<br>%{percent}<extra></extra>',
    showlegend=False
))

# Add center text
fig.add_annotation(
    text=f"<b>{region_name}</b><br>{total}",
    x=0.5, y=0.5,
    xref='paper', yref='paper',
    showarrow=False,
    font=dict(size=32, color='white', family='Arial'),
    align='center'
)

fig.update_layout(
    title={
        'text': f"{region_name} - Climate Commitments (2025)",
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 20, 'color': 'white'}
    },
    width=1200,
    height=1200,
    paper_bgcolor='#000000',
    plot_bgcolor='#000000',
    font=dict(color='white'),
    margin=dict(t=100, l=50, r=50, b=50)
)

fig.write_image(f"donut_{region_name.lower()}.png", width=1400, height=1400, scale=2)
fig.show()
print(f"Saved: donut_{region_name.lower()}.png")
```
