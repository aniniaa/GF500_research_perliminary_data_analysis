## Waterfall and sankey


```python
df24 = pd.read_csv(r"C:\Users\AniyaBagheri\OneDrive - Climate Impact\comparative analysis\gf5002024.csv", encoding="latin1")
df25 = pd.read_csv(
    r"C:\Users\AniyaBagheri\OneDrive - Climate Impact\comparative analysis\gf500 2026.csv",
    encoding="latin1",header=2
)
```

```python
print(df24.columns.tolist())
print(df25.columns.tolist())
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# --- Standardize company names ---
df24['Company_clean'] = df24['Company Name'].str.upper().str.strip()
df25['Company_clean'] = df25['Company Name'].str.upper().str.strip()

# --- Map statuses ---
def map_status_2024(row):
    if row['SBT (Y = yes, N=no)'] == 'N':
        return 'No SBT'
    if row['Target status - NT'] == 'T':
        return 'Target Set'
    if row['Target status - NT'] == 'C':
        return 'Committed'
    if row['Net zero committed (SBTi)'] == 'Y':
        return 'Net Zero Committed'
    return 'Other'

def map_status_2025(row):
    if row['Near Term SBT (Y = yes, N=no)'] == 'N':
        status = 'No SBT'
    elif row['Target status (T = target set, C = committed, N=No, R=Removed)'] == 'T':
        status = 'Target Set'
    elif row['Target status (T = target set, C = committed, N=No, R=Removed)'] == 'C':
        status = 'Committed'
    else:
        status = 'Other'
    if row['Net zero committed (SBTi) (Y = yes, N = no)'] == 'Y':
        status = 'Net Zero Committed'
    elif row['SBTi NZ Target Status'] == 'Approved':
        status = 'SBTi Approved'
    return status

df24['SBT_status_24'] = df24.apply(map_status_2024, axis=1)
df25['SBT_status_25'] = df25.apply(map_status_2025, axis=1)

# --- Compute sets ---
set24 = set(df24['Company_clean'])
set25 = set(df25['Company_clean'])
remained = set24 & set25
left = set24 - set25
entered = set25 - set24

# --- Merge for Sankey ---
trans = df24[df24['Company_clean'].isin(remained)].merge(
    df25[df25['Company_clean'].isin(remained)],
    on='Company_clean'
)

# --- Sankey data ---
status_labels = pd.concat([trans['SBT_status_24'], trans['SBT_status_25']]).unique()
status_map = {s:i for i,s in enumerate(status_labels)}

sankey_sources = []
sankey_targets = []
sankey_values = []

for s24, s25, count in trans.groupby(['SBT_status_24', 'SBT_status_25']).size().reset_index(name='count').values:
    sankey_sources.append(status_map[s24])
    sankey_targets.append(status_map[s25])
    sankey_values.append(count)

# --- Plot combined figure ---
fig = make_subplots(rows=1, cols=2, subplot_titles=("GF500 Waterfall 2024->2025", "SBTi Status Transitions"),
                    specs=[[{"type":"waterfall"}, {"type":"sankey"}]])

# Waterfall
fig.add_trace(go.Waterfall(
    measure=["relative", "relative", "relative", "total"],
    x=["Left 24->25", "Remained", "Entered 25", "Total 25"],
    y=[-len(left), len(remained), len(entered), len(df25)],
    text=[len(left), len(remained), len(entered), len(df25)],
    textposition="outside"
), row=1, col=1)

# Sankey
fig.add_trace(go.Sankey(
    node=dict(pad=15, thickness=20, label=list(status_map.keys())),
    link=dict(source=sankey_sources, target=sankey_targets, value=sankey_values)
), row=1, col=2)

fig.update_layout(title_text="GF500 2024->2025: Company Counts and SBTi Status Transitions", font_size=10)
fig.show()

```
